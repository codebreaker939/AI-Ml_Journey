```python
import json
```

```python
#loading the data

def load_data(filename): 
    with open (filename ,"r") as f:
        data = json.load(f)
    return data
```

```python
data = load_data("product_data.json")
print( data)
print(type(data)) 
```

```python
def cln_data(data):
    # 1. txt to num of rating and cleaning it
    txt_to_num = {'one': 1, 'two': 2, 'three': 3, 'four': 4, 'five': 5}
    for user in data:
        raw_rating = str(user.get('rating', '')).strip().lower()
        if raw_rating in txt_to_num:
            user['rating'] = txt_to_num[raw_rating]

    # 2. handle missing value
    for user in data:
        raw_age = user.get("age")
        if raw_age is None:
            user['age'] = 0

    # 3. handling duplicates via deduplication
    cleaned_data = []
    unique_user = set()
    for user in data:
        name = user.get('name')
        if name in unique_user:
            continue
        unique_user.add(user['name'])
        cleaned_data.append(user)

    return cleaned_data
```

```python
newdata = cln_data(data)
```

```python
# getting insights from data 

def insights(data):

    # avg rating
    tot_rating = 0
    for user in data:
        tot_rating+= float(user['rating'])

    print(f"total avg ratin is {tot_rating/len(data)}")

    # user % for poor rating
    poor_rating =0
    for user in data:
        if(float(user['rating'])<3):
            poor_rating+=1
    print(poor_rating/len(newdata)*100)
        
```

```python
insights(newdata)
```

```python
def recommend (data):

    # recommendation 
    reccom=[]
    for user in data:
        curr_reccom={}
        curr_reccom['name']=user['name']
        if(float(user['rating'])<4):
                  curr_reccom['brand']='apple'
        else : 
                  curr_reccom['brand']='samsung'

        reccom.append(curr_reccom)

    return reccom
```

```python
recommend(newdata)
```
